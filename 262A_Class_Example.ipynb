```python
path = "/content/drive/MyDrive/Library_Usage.csv"

import pandas as pd

df = pd.read_csv(path)

df.head()
```

```python
df["Patron Type Definition"].value_counts()
```

```python
df["Total Checkouts"].value_counts()
```

```python
import matplotlib.pyplot as plt

df["Year Patron Registered"].value_counts().plot(
    kind="pie",
)

plt.title("Patron Registrations by Year")
plt.show()
```

```python
import matplotlib.pyplot as plt

df["Year Patron Registered"].value_counts().plot(
    kind="bar",
)

plt.title("Patron Registrations by Year")
plt.show()
```

```python
import matplotlib.pyplot as plt

df["Year Patron Registered"].value_counts().plot(
    kind="bar",
    color="pink",
    figsize=(11,8),
)

plt.title("Patron Registrations by Year")
plt.show()
```

```python
import matplotlib.pyplot as plt

df["Circulation Active Month"].value_counts().plot(
    kind="bar",
    color="pink",
    figsize=(11,8),
)

plt.title("Library Circulation Activity by Month")
plt.ylabel("# of Circulations")
plt.xlabel("Month")
plt.show()
```

```python
month_order = [
    "January", "February", "March", "April", "May", "June", "July", "August", "September", "October", "November", "December"
]

df["Circulation Active Month"].value_counts().reindex(month_order).plot(
    kind="bar",
    color="pink",
    edgecolor="blue",
    figsize=(11,8),
)

plt.title("Library Circulation Activity by Month")
plt.ylabel("# of Circulations")
plt.xlabel("Month")
plt.show()
```

```python
df.isna().sum()
```

```python
df["Circulation Active Month Cleaned"] = df["Circulation Active Month"].fillna("Not Captured")

df.isna().sum()
```

```python
month_order = [
    "January", "February", "March", "April", "May", "June", "July", "August", "September", "October", "November", "December", "Not Captured"
]

df["Circulation Active Month Cleaned"].value_counts().reindex(month_order).plot(
    kind="bar",
    color="yellow",
    edgecolor="pink"
)

plt.title("Library Circulation Activity by Month")
plt.ylabel("# of Circulations")
plt.xlabel("Month")
plt.show()
```

```python
age_order = [
    "0 to 9 years", "10 to 19 years", "20 to 24 years", "25 to 34 years", "35 to 44 years", "45 to 54 years", "55 to 59 years", "60 to 64 years", "65 to 74 years", "75 years and over", "Not Captured"
]

df["Age Range Cleaned"].value_counts().reindex(age_order).plot(
    kind="bar",
    color="pink",
    edgecolor="brown",
    figsize=(11,8)
)

plt.title("Library Patron Age Range")
plt.show()
```

```python
df["Age Range Cleaned"] = df["Age Range"].fillna("Not Captured")

df.isna().sum()
```
